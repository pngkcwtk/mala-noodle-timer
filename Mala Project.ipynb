```python
# import time อันนี้ต้องเรียกใช้ Module
from tkinter import * # * เอามาทุกตัวที่ต้องการ, ถ้าใส่แค่ชื่อคำสั่ง ก็จะใช้ได้แค่คำสั่งนั้น, ไม่ต้องเรียกใช้ Module
```

```python
global label, window
noodle = input("Enter your noodle :")
```

```python
def timer_noodles(seconds): # ใส่เวลาหน่วยวินาที
    """Countdown"""
    if seconds > -1 :
        mins, secs = divmod(seconds, 60)  # แบ่งวินาทีเป็นนาทีและวินาที
        timer = f"{mins:02}:{secs:02}"  # ฟอร์แมตเวลาเป็น MM:SS
        print(timer, end = '\r')  # แสดงเวลาในบรรทัดเดียว
        
        #time.sleep(1)  # หยุดการทำงาน 1 วินาที
        #seconds -= 1
        
        label.config(text=timer)
        window.after(1000, timer_noodles, seconds - 1)
    else :
        label.config(text="!!Finish!!\nพร้อมรับประทานแล้ว :D")
        print(f"\n{noodle}\n!!Finish!!\nพร้อมรับประทานแล้ว :D")
```

```python
def main(noodle):
    """input noodle"""
    
    if noodle == "Sweet Potato Noodles" or noodle == "Glass Noodles" :
        timer_noodles(420)
    elif noodle == "Konjac Noodles" :
        timer_noodles(300)
    elif noodle == "Fern Root Noodles" :
        timer_noodles(600)
    else : # เส้นอื่น ๆ หรือจับเวลาอื่น ๆ
        mins, secs = int(input()), int(input())
        seconds = (mins*60) + secs
        timer_noodles(seconds)
```

```python
window = Tk()

# ลักษณะภายนอก UI Outside
window.title("Timer") # ชื่อกล่อง ซ้ายบน
window.geometry("450x250") # size
window.minsize(25, 20) #ย่อได้มากสุดเท่าไร
window.eval("tk::PlaceWindow . center") # ทำให้ window โผล่ตรงกลางหน้าจอ
# window.resizable(0, 0) ไม่ต้องการให้ย่อขนาด window ได้ 0, 0 ย่อไม่ได้เลย

# ใส่ข้อมูล การประมวลผล รับค่า
frame = Frame(window)
frame.pack(expand=True)

name = Label(frame, text=noodle, font="time15")
label = Label(frame, font="time15")

name.pack()
label.pack() # ทำให้แสดงผลบน window

main(noodle)

window.mainloop() # เปิด window
```
